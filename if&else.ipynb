# if else and elif.

* elif is used for multiple if condition.

```python
num = int(input("Enter a number:"))
if num == 0:
    print('number is zero')
elif num > 0:
    print('number is positive')
else:
    print('number is negative')
```

# nested if else

```python
num = -56
if num > 0:
    if num % 2 == 0:
        print('even')
    else:
        print('odd')
else:
    print('number is negative')
```

# Check if a year is Leapyear or not.

```python
year = int(input('enter year'))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print('This is a Leap Year')
else:
    print('This is not a Leap year')
```

# To check vowel is present or not in character


```python
user = input('enter a word')
if user in 'aeiouAEIOU':
    print('vowel')
else:
    print('consonant')
```

* another method to write this code:- 

```python
user = input('enter a word')
output = 'vowel' if user in 'aeiouAEIOU' else 'consonant'
print(output)
```

# fare ticket price check, if age is 18+ then
20 for adult
10 for childrens
and extra 5 Rs discount. if today is saturday


```python
age = int(input('enter your age'))
day = str(input('enter day of the week'))
price = 20
if age > 20:
    price = 20
else:
    price = 10

if day == 'saturday':
    price -= 5
print('final price is ',price)
```
