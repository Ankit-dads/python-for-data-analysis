# if else:

```python
name = input("What is your name?")
if name == 'Ankit':
    print("my name is Ankit")
else:
    print("I don't know my name.")
```

```python
age = int(input("What is your age?"))
if age >= 18:
    print("you are eligible to vote.")
else:
    print('You are not eligible to vote.')
```

```python
num = int(input('enter number'))
if num % 2 == 0:
    print("number is evern")
else:
    print("number is odd")
```

# if else and elif.

* elif is used for multiple if condition.

```python
num = int(input("Enter a number:"))
if num == 0:
    print('number is zero')
elif num > 0:
    print('number is positive')
else:
    print('number is negative')
```

# nested if else

```python
num = -56
if num > 0:
    if num % 2 == 0:
        print('even')
    else:
        print('odd')
else:
    print('number is negative')
```

```python
wages =  5000
credit = int(input('enter credit score'))
if wages >= 4800:
    if credit >= 70:
        print('can take personal loan')
    else:
        print('cannot take personal loan')
else:
    print('salary is not sufficient to take personal loan')
```
