## diff between remove and discard

* if a value is not present in the set then by using remove fn it will give error but if we use discard then it will remove the value if it is present in the set, but if it is not then it will not give any error

```python
set.remove(34)
```

```python
set.discard('Niharika')
```

# Dictionary ----->


* key value pairs {a:b}
* mutable
* combination of both keys and values while keys are immutable and values are mutable
* {keys(immutable):values(mutable)}
* keys always be unique

```python
my_dict = {'name':'Ankit kashyap',
           'Age':23,
           'country':'India',
           'Religion':"Hindu"
           }
print(my_dict)
print(type(my_dict))
```

```python

my_dict['Religion']
```

* how to update any value in dict.

```python
my_dict['Religion'] = 'Sanatani'
my_dict
```

* How to add any new key value pair in dict.

```python
my_dict['Profession'] = 'Trader'
my_dict
```

```python
my_dict.keys()
```

```python
my_dict.values()
```

* items() will give the pair of keys & values

```python
my_dict.items()
```

* how to update/add multiple key value pairs in dict simultaneously

```python
my_dict.update({'Married':False,"Income":'7 million $','Religion':'Hindu'})
my_dict
```

* pop will do the same fn in dict 

```python
my_dict.pop('Married')
```
