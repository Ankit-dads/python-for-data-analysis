# Array creation with Random method

```python
# np.random() give uniform distribuion of random vales between 0 and 1
random = np.random.rand(3,4)
random
```

```python
# np.random.randn creates an array of the given shape with random values from a standard normal distribution (0,1)
random2 = np.random.randn(5,6)
print(random2)
```

```python
# np.random.randint(start, stop,size)
# it will give values between 1 to 10 and it will give 12 values
random3 = np.random.randint(1,10,12)
random3
```

* we can also create matrix with np.random.randint

```python
mat = np.random.randint(2,10,(5,6)) # it will give values between 2 to 10 and it will give 5 rows and 6 columns
mat
```

```python
a = np.random.randint(0,25,26) # generate a ranodom integer between 0 and 25 or the range we will pass
print(a)
```

# diff btw random.randint and random_integers

```python
# np.random.random_integer(start,stop,size) 
# import warnings
# warnings.filterwarnings("ignore")

b = np.random.random_integers(2,4,9) # start and end point is inclusive in this function this is the diff between random_integers and randint
b
```

```python
mat1 = np.random.random_integers(1,10,(5,4)) # can also generate a matrix using random_integers by passing shape of the matrix
mat1
```
