* indexing and slicing with n-dimensional array

```python
a = [1,2,3,4]
b = [4,5,6,7]
c = [5,7,8,9]
data1 = np.array([a,b,c]) # IN python data structure there is also a method like that called packing and unpacking of list.
```

```python
data1
```

* now we have a matrix of 4 by 3 if we want to extract a single element from it or a group of element then we perform slicing to it so that with the given indices the desired data is extracted

```python
# if we want to get the middle dataa like 9 so the value/indexing for row is 2 and for column is 3 so the indexing will be like 2:3
data1[2,3] #[row,column]
```

```python
# if we want a range of data then we give indexing to range or row and column like [a:b,c:d] where a:b is row indexing and c:d is column indexing the our data is extracted or sliced
data1[1:3,1:3] # here the value of row and column is n-1 means 1:4 means 1:3 ONLY
```

* vstack in numpy
 np.vstack is a function provided by the NumPy library in Python for verticalstacking of arrays. Vertical stacking, or "vstack," refers to stacking arrays ontop of each other along the rows. It combines arrays by adding rows,effectively increasing the number of rows in the resulting array.

```python
array1 = np.array([1,2,3])
array2 = np.array([4,5,6])
verticalstack = np.vstack((array1,array2))
```

```python
verticalstack # now we can perform various other fun on it like shape reshape etc.
```

* column stack :->
Column stacking, often referred to as "column_stack," is the process ofstacking arrays side by side along columns to create a new array. This operation is useful when you want to combine multiple 1D arrays or lists into a 2D array, where each original array becomes a column in the resulting 2Darray.

```python
np.column_stack([array1,array2])
```

*  Horizontal stacking, often referred to as hstack, is the process of combiningmultiple arrays horizontally along the columns. This means that the arrays arestacked side by side, effectively extending the number of columns in theresulting array.

```python
np.hstack([array1,array2])
```
